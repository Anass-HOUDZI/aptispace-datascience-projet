# Chapitre 1 — Acquisition des Données

Ce notebook couvre l'ingestion du fichier brut `marketing_campaign.csv`, l'audit mémoire initial et la première inspection de qualité.

```python
import os, sys, warnings
warnings.filterwarnings('ignore')
sys.path.insert(0, os.path.abspath('../../src'))

import pandas as pd
import numpy as np

RAW_PATH = os.path.abspath('../../data/raw/marketing_campaign.csv')

# Lecture avec engine C et separateur tabulation
df = pd.read_csv(RAW_PATH, sep='\t', engine='c')

print(f'Dimensions brutes : {df.shape[0]} lignes x {df.shape[1]} colonnes')
print(f'Memoire initiale  : {df.memory_usage(deep=True).sum() / 1024:.1f} KB')
```

## 1.1 Apercu des Premieres Lignes

```python
df.head(5)
```

## 1.2 Types et Valeurs Manquantes

```python
info = pd.DataFrame({
    'dtype'  : df.dtypes,
    'nulls'  : df.isnull().sum(),
    'null_%' : (df.isnull().sum() / len(df) * 100).round(2),
    'unique' : df.nunique(),
})
print(info.to_string())
```

## 1.3 Downcasting — Optimisation Memoire

```python
mem_avant = df.memory_usage(deep=True).sum() / 1024

for col in df.select_dtypes(include=['int64']).columns:
    df[col] = pd.to_numeric(df[col], downcast='integer')
for col in df.select_dtypes(include=['float64']).columns:
    df[col] = pd.to_numeric(df[col], downcast='float')

mem_apres = df.memory_usage(deep=True).sum() / 1024
gain = (1 - mem_apres / mem_avant) * 100

print(f'Avant  : {mem_avant:.1f} KB')
print(f'Apres  : {mem_apres:.1f} KB')
print(f'Gain   : -{gain:.0f}% memoire')
```

## 1.4 Statistiques Descriptives Rapides

```python
print('=== Variables numeriques ===')
print(df.describe().T[['mean','std','min','max']].round(2).to_string())
print()
print('=== Variables categorielles ===')
for col in df.select_dtypes(include='object').columns:
    print(f'{col:25s} | {df[col].nunique()} modalites | ex: {df[col].unique()[:3]}')
```

## 1.5 Audit Qualite — Anomalies Detectees

| Anomalie | Variable | Valeur | Action |
|----------|----------|--------|---------|
| Valeurs manquantes MNAR | `Income` | 24 NaN | Flag binaire + imputation mediane strate |
| Outliers physiques | `Year_Birth` | 1893, 1900 | Masquage (age > 100 ans impossible) |
| Modalites non standards | `Marital_Status` | YOLO, Absurd, Alone | Harmonisation -> Other |
| Format date texte | `Dt_Customer` | DD-MM-YYYY | Conversion datetime64 + calcul Customer_Days |

```python
# Verification des anomalies
print('Valeurs manquantes sur Income :', df['Income'].isnull().sum())
print('Annees de naissance anormales :', df[df['Year_Birth'] < 1920]['Year_Birth'].tolist())
print('Modalites Marital_Status :', df['Marital_Status'].value_counts().to_dict())
print('Type Dt_Customer :', df['Dt_Customer'].dtype, '| Ex:', df['Dt_Customer'].iloc[0])
```
